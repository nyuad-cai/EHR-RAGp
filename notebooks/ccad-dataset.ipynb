```python
import torch
import torch.nn as nn
def get_pretrained_weights( model: nn.Module, ckpt_path: str) -> None:
    # 1) load Lightning checkpoint
    sd = torch.load(ckpt_path, map_location="cpu", weights_only=False)["state_dict"]

    # 2) figure out which HF model this is (bert, roberta, big_bird, ...)
    model_type = getattr(getattr(model, "config", None), "model_type", "").lower()

    # 3) map model_type -> inner module name used during pretraining
    #    (i.e. the thing after "backbone.")
    inner_name_map = {
        "bert": "bert",
        "big_bird": "bert",       # BigBirdForMaskedLM.bert
        "roberta": "roberta",
        "longformer": "longformer",
        "roformer": "roformer",
        "modernbert": "model",
    }

    inner = inner_name_map.get(model_type)
    if inner is None:
        print(f"[get_pretrained_weights] Unknown model_type='{model_type}', "
              f"falling back to stripping only 'backbone.'")
        main_prefix = "backbone."
    else:
        main_prefix = f"backbone.{inner}."

    # 4) prefixes to drop (MLM heads etc.)
    drop_prefixes = (
        "backbone.cls.",        # BERT/Roberta/Longformer/BigBird/RoFormer MLM head
        "backbone.head.",       # ModernBertForMaskedLM head
        "backbone.lm_head.",    # other possible naming
        "backbone.predictions.",
    )

    remapped = {}
    for k, v in sd.items():
        # skip MLM heads
        if any(k.startswith(dp) for dp in drop_prefixes):
            continue

        # keep only backbone.<inner>.* and strip that prefix
        if k.startswith(main_prefix):
            new_k = k[len(main_prefix):]  # e.g. backbone.bert.embeddings... -> embeddings...
            remapped[new_k] = v
        else:
            # everything else (ehr_embeddings.*, optimizer states, etc.) is ignored
            continue

    missing, unexpected = model.load_state_dict(remapped, strict=False)
    print("weights loaded successfully!")
    print("missing keys:", missing)
    print("+" * 50)
    print("unexpected keys:", unexpected)
    
```

```python
from transformers import CONFIG_MAPPING, MODEL_FOR_MASKED_LM_MAPPING, MODEL_MAPPING
```

```python
def get_config_and_model_cls(model_type: str, mode: str = 'mlm'):
    if model_type not in CONFIG_MAPPING:
        raise ValueError(f"Unknown model_type: {model_type}")
    config_cls = CONFIG_MAPPING[model_type]   # e.g. BertConfig, ModernBertConfig

    if mode == 'mlm':  
        if config_cls not in MODEL_FOR_MASKED_LM_MAPPING:
            raise ValueError(f"No MaskedLM model registered for config: {config_cls}")
        model_cls = MODEL_FOR_MASKED_LM_MAPPING[config_cls]  # e.g. BertForMaskedLM
    else:
        if config_cls not in MODEL_MAPPING:
            raise ValueError(f"No base model registered for config: {config_cls}")
        model_cls = MODEL_MAPPING[config_cls]

    return config_cls, model_cls
```
